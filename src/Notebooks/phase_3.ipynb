## Advance Operations

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Que.1 Zomato Sales Analysis

```python
# Data structure: [restaurant_it, 2021, 2022, 2023, 2024]
sales_data = np.array([

[1, 150000, 180000, 140000, 250000], # Paradise Biryani

[2, 120000, 170000, 160000, 190000], # Beijing Bites

[3, 200000, 230000, 180000, 300000], # Pizza Hub

[4, 180000, 150000, 240000, 270000], # Burger Point

[5, 160000, 185000, 205000, 220000]  # Chai Point

])

print("==== Zomato sales analysis ====")
print("\n Sales data shape:", sales_data.shape)

print("\n Sample data for 1st three restaurant: \n", sales_data[0:3])
```

#### Total sales per year

```python
sales_per_year = np.sum(sales_data[:,1:], axis=0)
# print("Sales per year:\n",sales_per_year)
print(f"Sales in year 2021:\n{sales_per_year[0]}")
print(f"Sales in year 2022:\n{sales_per_year[1]}")
print(f"Sales in year 2023:\n{sales_per_year[2]}")
print(f"Sales in year 2024:\n{sales_per_year[3]}")
```

#### Total Sales(2021-2024) Item wise

```python
Sales_per_restaurant = np.sum(sales_data[:,1:], axis = 1)
print(f"Sales per restaurant:\n{Sales_per_restaurant}")
print("\nFrom 2021 to 2024...\n")
restaurants = ["Paradise Biryani", "Beijing Bites", "Pizza Hub", "Burger Point", "Chai Point"]


for restaurant, sales in zip(restaurants ,Sales_per_restaurant):
    print(f"Total sales for {restaurant}: {sales}")
```

#### Min and Max sales per restaurant

```python
min_sales = np.min(sales_data[:,1:], axis = 1)
print(min_sales)

max_sales = np.max(sales_data[:,1:], axis = 1)
print(max_sales)

years = [2021, 2022, 2023, 20224]
for restaurant,min_sale,max_sale,year in zip(restaurants, min_sales, max_sales, years):
    print(f"For {restaurant} → Min Sales: ₹{min_sale}, Max Sales: ₹{max_sale}")
```

#### Average sales per restaurant

```python
avg_Sales_per_restaurant = np.sum(sales_data[:,1:], axis = 1)/len(years)
avg_sales = np.mean(sales_data[:,1:], axis = 1)
print(f"Sales per restaurant:\n{avg_Sales_per_restaurant}")
print("\nFrom 2021 to 2024...\n")
restaurants = ["Paradise Biryani", "Beijing Bites", "Pizza Hub", "Burger Point", "Chai Point"]

print("Using len() function\n")

for restaurant, sales in zip(restaurants ,avg_Sales_per_restaurant):
    print(f"Avg. sales for {restaurant}:-  {sales}")

print("\nUsing mean() function\n")

for restaurant, sales in zip(restaurants ,avg_sales):
    print(f"Avg. sales for {restaurant}:-  {sales}")
```

#### Cumulative Sum

```python
cumulative_sales = np.cumsum(sales_data[:, 1:], axis = 1)
print(cumulative_sales)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumulative_sales, axis=0))
plt.title("Average cumulative sales accross all restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### Vectors

#### Vectors Addition

```python
vector_1 = np.array([4,6,3])
vector_2 = np.array([3,8,6])
print(f"Vector addition:\n{vector_1} + {vector_2} = {vector_1 + vector_2}\n")
print(f"Vector subtraction:\n{vector_1} - {vector_2} = {vector_1 - vector_2}")

print("\nvector_1 + vector_2 = vector_2 + vector_1")
print("\nBut,  vector_1 - vector_2 != vector_2 - vector_1\nInfact, vector_1 - vector_2 = -(vector_2 - vector_1)")
```

#### Vector Multiplication
Vectors Multiplication, Dot Product and Cross Product 

```python
print(f"Vector Multiplication:\n{vector_1} * {vector_2} = {vector_1*vector_2}")

print("\nDot Product:")
print(f"{vector_1} . {vector_2} = {np.dot(vector_1,vector_2)}   >> which is equal to sum of all elements of array vector_1 * vector_2 ie. {vector_1*vector_2}")

print("\nCross Product:")
print(f"{np.cross(vector_1,vector_2)}")

print("\nDot product is a mathematical number\nBut Cross product is another Vector\n\nvector_1 . vector_2 = vector_2 . vector_1\nBut, vector_1 X vector_2 = -(vecotr_2 X vector_1)")
```

#### Magnitude of Vectors and Angle b/w Vectors

```python
print("Magnitude of vectors:\n")
print(f"Magnitude of vector 1: {round(np.linalg.norm(vector_1), 2)}")
print(f"Magnitude of vector 2: {round(np.linalg.norm(vector_2))}\n")

print("\nAngle between two vectors:\n")
angle = round(np.arccos((np.dot(vector_1,vector_2)) / (np.linalg.norm(vector_1) * (np.linalg.norm(vector_2)))),4)
print(f"Angle between vectors {vector_1} and {vector_2} in radians is {angle}")
print(f"Angle between vectors {vector_1} and {vector_2} in degrees is {np.degrees(angle)}")
```
