## Numpy >> arrays and basics
Introduction to arrays

```python
import numpy as np
import time
```

### Creating array

```python
arr_1D = np.array([1,2,3,4])
print(f"1D array: {arr_1D}")

arr_2D = np.array([[1,2,3],[4,5,6]])
print(f"2D array: {arr_2D}")
```

#### List vs Numpy array

```python
print("Multiplication with scaler...\n")

py_list = [1,2,3]
print(f'''Python list multiplication example:
      {py_list}*2 = {py_list*2}''')
print("")

#element vise multiplication
np_array = np.array([1,2,3])
print(f'''Numpy array multiplication example:
      {np_array}*2 = {np_array*2}''')


start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operation time: ", time.time() - start)
```

#### Creating array from scratch

```python
zeros = np.zeros((3,4))          # zeros array
print(f"Zeros array: \n{zeros}")

ones = np.ones((2,3))            # ones array
print(f"Ones array: \n{ones}")

full = np.full((2,2),7)          # full array
print(f"Full array: \n{full}")

random = np.random.random((2,3)) # random array
print(f"Random array: \n{random}")

sequence = np.arange(0,11,2)     # array with the help of range
print(f"Sequence array: \n{sequence}")
```

### Vector, Matrix and Tensor

```python
vector = np.array([1, 2, 3])                            # 1D
print(f"\nVector:\n{vector}")

matrix = np.array(([                                    # 2D
    [1, 2, 3],
    [4, 5, 6]
]))
print(f"\nMatrix:\n{matrix}")

tensor = np.array([[[1,2],[3,4]],
                   [[5,6], [7,8]]])       # 3D
print(f"\nTensor:\n{tensor}")
```

### Array properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print(f"Array:\n{arr}\n")
print(f"Shape: {arr.shape}")
print(f"Dimension: {arr.ndim}")
print(f"Size: {arr.size}")
print(f"Dtype: {arr.dtype}")
```

#### Methods on arrays

```python
arr = np.arange(12)
print(f"Original array:\n{arr}\n")

reshaped_arr = arr.reshape((3,4))
print(f"Reshaped array:\n{reshaped_arr}\n")

# it returns copy
flattened_arr = reshaped_arr.flatten()
print(f"Flattened array:\n{flattened_arr}\n")

# it returns view, instead of copy
raveled = reshaped_arr.ravel()
print(f"Raveled array:\n{raveled}\n")

transpose_1 = reshaped_arr.transpose()
transpose_2 = reshaped_arr.T   # it is an attribute (@property method)
print(f"Transposed array:\n{transpose_1}\n{transpose_2}\n")
```
